```python
import pandas as pd
```

```python
df_ground_truth = pd.read_excel('ground_truth_full.xlsx').drop(columns='Language')
df_title = pd.read_excel('headlines_dataset.xlsx')[['Filename', 'Headline_GER']]

# Merge using left_on and right_on
ground_truth_title = pd.merge(
    left = df_ground_truth,
    right = df_title,
    left_on='Filename',
    right_on='Filename',
    how='inner'
)

ground_truth_title.to_excel('ground_truth_title.xlsx', index=False)
```

```python
df_LLM = pd.read_excel('03_finale_titelannotationen.xlsx')

cols = [
    "Economic",
    "Capacity and resources",
    "Morality",
    "Fairness and equality",
    "Legality, constitutionality and jurisprudence",
    "Policy prescription and evaluation",
    "Crime and punishment",
    "Security and defense",
    "Health and safety",
    "Quality of life",
    "Cultural identity",
    "Public opinion",
    "Political",
    "External regulation and reputation"
    ]

df_LLM804 = df_LLM[df_LLM[cols].eq(1).any(axis=1)].drop(columns=['Language'])

df_LLM804.to_excel('match_LLM804.xlsx', index=False)
```

```python
# 1. Load the Excel files
ground_truth_title = pd.read_excel('ground_truth_title.xlsx')
df_filter = pd.read_excel('match_LLM804.xlsx')

# 2. Extract valid IDs from the 'article id' column
valid_ids = df_filter['article id']

# 3. Filter ground truth where 'Filename' matches valid IDs
filtered_df = ground_truth_title[ground_truth_title['Filename'].isin(valid_ids)].drop(columns='Filename')

# 4. Save to a new file
filtered_df.to_excel('training_data/ground_truth_filtered.xlsx', index=False)
```
