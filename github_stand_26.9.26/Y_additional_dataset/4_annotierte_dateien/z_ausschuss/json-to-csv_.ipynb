```python
import json
from pathlib import Path
import pandas as pd


def convert_json_to_csv(
    json_path: str,
    csv_path: str,
    sep: str = ";",
    encoding: str = "utf-8-sig",
) -> pd.DataFrame:
    path = Path(json_path)
    if not path.exists():
        raise FileNotFoundError(
            f"Die Datei '{json_path}' wurde nicht gefunden."
        )

    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    if isinstance(data, dict):
        list_keys = [k for k, v in data.items() if isinstance(v, list)]
        if len(list_keys) == 1:
            data = data[list_keys[0]]
        else:
            data = [data]

    df = pd.json_normalize(data)
    df.to_csv(csv_path, index=False, sep=sep, encoding=encoding)

    print("✓ Konvertierung erfolgreich!")
    print(f"  Quelle: {json_path}")
    print(f"  Ziel:   {csv_path}")
    print(f"  Zeilen: {len(df)}, Spalten: {len(df.columns)}")

    return df


# Pfade festlegen
INPUT_JSON = "chn.json"
OUTPUT_CSV = "chn_annotated_gemini.csv"

# Konvertierung direkt ausführen
df_result = convert_json_to_csv(
    json_path=INPUT_JSON,
    csv_path=OUTPUT_CSV,
    sep=";",
    encoding="utf-8-sig",
)

# Vorschau im Notebook anzeigen
df_result.head()
```
