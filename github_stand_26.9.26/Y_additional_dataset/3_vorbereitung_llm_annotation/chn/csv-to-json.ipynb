```python
import pandas as pd

input_csv = 'chn_headlines_dataset_full_LLM.csv'
output_json = 'chn_headlines_dataset_full_LLM.json'

# CSV mit Semikolon-Trennzeichen einlesen
df = pd.read_csv(input_csv, sep=';', encoding='utf-8-sig')

# Als lesbares JSON im Datensatz-Format (List of Objects) speichern
df.to_json(output_json, orient='records', force_ascii=False, indent=2)

print(f"Erfolgreich {len(df)} Einträge in '{output_json}' gespeichert.")
```
