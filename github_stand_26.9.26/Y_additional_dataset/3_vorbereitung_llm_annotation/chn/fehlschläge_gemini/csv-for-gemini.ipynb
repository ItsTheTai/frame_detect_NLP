```python
import csv
from io import StringIO

# Pfad zur Eingabedatei
input_file = "chn_headlines_dataset_full_LLM.csv"

# Ausgabedatei (optional, falls du die Datei speichern möchtest)
output_file = "gemini_compatible_chn_headlines_dataset_full_LLM.csv"

# 1. Datei einlesen und BOM (﻿) entfernen
with open(input_file, "r", encoding="utf-8-sig") as f:
    content = f.read()

# 2. Semikolon durch Komma ersetzen (für standard CSV)
content = content.replace(";", ",")

# 3. In eine neue Datei schreiben (oder direkt ausgeben)
with open(output_file, "w", encoding="utf-8") as f:
    f.write(content)

# 4. Optional: Inhalt direkt als Text ausgeben (für Copy-Paste)
print("--- FERTIGES FORMAT FÜR GEMINI ---")
print(content)
```
