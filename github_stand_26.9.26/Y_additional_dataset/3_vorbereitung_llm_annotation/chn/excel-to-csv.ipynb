```python
from pathlib import Path
import pandas as pd


def convert_excel_to_csv(folder_path: Path):
    # .glob() sucht NUR direkt im angegebenen Ordner (keine Unterordner)
    excel_files = list(folder_path.glob("*.xlsx")) + list(
        folder_path.glob("*.xls")
    )

    if not excel_files:
        print(f"Keine Excel-Dateien in '{folder_path}' gefunden.")
        return

    for file in excel_files:
        # Temporäre Excel-Sperrdateien (fangen mit ~$ an) überspringen
        if file.name.startswith("~$"):
            continue

        try:
            # Liest alle Tabellenblätter der Datei ein
            sheets = pd.read_excel(file, sheet_name=None)

            for sheet_name, df in sheets.items():
                # Bei einem Blatt bleibt der Dateiname gleich, bei mehreren wird der Blattname angehängt
                if len(sheets) == 1:
                    output_path = folder_path / f"{file.stem}.csv"
                else:
                    output_path = folder_path / f"{file.stem}_{sheet_name}.csv"

                # sep=";" als gängiges Trennzeichen
                # utf-8-sig sorgt für korrekte Umlaute in Excel
                df.to_csv(
                    output_path, index=False, sep=";", encoding="utf-8-sig"
                )
                print(f"Erfolgreich konvertiert: {output_path.name}")

        except Exception as e:
            print(f"Fehler bei Datei '{file.name}': {e}")


if __name__ == "__main__":
    # Ermittelt dynamisch den Ordner, in dem dieses Skript liegt
    SKRIPT_ORDNER = Path.cwd()

    convert_excel_to_csv(SKRIPT_ORDNER)
```
