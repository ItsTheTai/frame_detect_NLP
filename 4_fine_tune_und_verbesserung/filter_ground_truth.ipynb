```python
df_ground_truth = pd.read_excel('frame_detect_NLP/4_fine_tune_und_verbesserung/ground_truth_full.xlsx').drop(columns='Language')
df_title = pd.read_excel('frame_detect_NLP/4_fine_tune_und_verbesserung/headlines_dataset.xlsx')[['Filename', 'Headline_GER', 'Language']]

# Merge using left_on and right_on
ground_truth_title = pd.merge(
    left = df_ground_truth,
    right = df_title,
    left_on='Filename',
    right_on='Filename',
    how='inner'
)

ground_truth_title.to_excel('frame_detect_NLP/4_fine_tune_und_verbesserung/ground_truth_title.xlsx', index=False)
```

```python
df_LLM = pd.read_excel('frame_detect_NLP/4_fine_tune_und_verbesserung/03_finale_titelannotationen.xlsx')

cols = [
    "Economic", # 0
    "Capacity and resources", # 1
    "Morality", # 2
    "Fairness and equality", # 3 F T
    "Legality, constitutionality and jurisprudence", # 4
    "Policy prescription and evaluation", # 5
    "Crime and punishment", # 6
    "Security and defense", # 7
    "Health and safety", # 8 T
    "Quality of life", # 9 F (raus bei Titel)
    "Cultural identity", # 9-10 F T
    "Public opinion", # 10-11 F
    "Political", # 11-12
    "External regulation and reputation" #12-13
    ]

df_LLM804 = df_LLM[df_LLM[cols].eq(1).any(axis=1)]#.drop(columns=['Language'])

df_LLM804.to_excel('frame_detect_NLP/4_fine_tune_und_verbesserung/match_LLM804.xlsx', index=False)
```

```python
# 1. Load the Excel files
ground_truth_title = pd.read_excel('frame_detect_NLP/4_fine_tune_und_verbesserung/ground_truth_title.xlsx')
df_filter = pd.read_excel('frame_detect_NLP/4_fine_tune_und_verbesserung/match_LLM804.xlsx')

# 2. Extract valid IDs from the 'article id' column
valid_ids = df_filter['article id']

# 3. Filter ground truth where 'Filename' matches valid IDs
filtered_df = ground_truth_title[ground_truth_title['Filename'].isin(valid_ids)].drop(columns='Filename')

# 4. Save to a new file
filtered_df.to_excel('frame_detect_NLP/4_fine_tune_und_verbesserung/4_training_data/ground_truth_filtered.xlsx', index=False)
```
