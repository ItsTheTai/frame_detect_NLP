```python
import math
import pandas as pd

# Konfiguration
input_file = "mfc_dataset_headlines_full_LLM.csv"
num_splits = 12

# CSV-Datei einlesen (Semikolon-getrennt, UTF-8-SIG für BOM)
df = pd.read_csv(input_file, sep=";", encoding="utf-8-sig")

# Zeilenanzahl pro Teildatei berechnen
chunk_size = math.ceil(len(df) / num_splits)

# Datei in 12 Teile splitten und speichern
for i in range(num_splits):
    start_idx = i * chunk_size
    end_idx = (i + 1) * chunk_size
    chunk = df.iloc[start_idx:end_idx]

    if chunk.empty:
        break

    # Dateiname nummeriert von 01 bis 12
    output_file = f"mfc_dataset_headlines_part_{i+1:02d}.csv"
    chunk.to_csv(output_file, sep=";", index=False, encoding="utf-8")
    print(
        f"Erstellt: {output_file} ({len(chunk)} Datenzeilen + Spaltenüberschrift)"
    )
```
