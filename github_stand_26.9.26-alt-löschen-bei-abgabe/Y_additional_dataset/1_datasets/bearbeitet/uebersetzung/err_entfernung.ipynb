```python
import pandas as pd

# 1. Excel-Datei einlesen
file_path = "manuell_ergaenzt_ground_truth_chinese_extension_semeval_with_titles_translated_partial_refined.xlsx"
df = pd.read_excel(file_path)

# 2. Zeilen filtern
# Variante A (Empfohlen): Entfernt exakte "err"-Einträge sowie "Error 500..." Serverfehler.
# So werden reguläre deutsche Wörter (z. B. "Anti-Terror" oder "recycelter") nicht versehentlich gelöscht.
is_error = (
    df['Headline_GER'].astype(str).str.strip().str.lower().isin(['err', 'error']) | 
    df['Headline_GER'].astype(str).str.startswith('Error')
)
df_clean = df[~is_error]

# Variante B: Falls du ausnahmslos jeden Text löschen möchtest, der die Buchstabenfolge "err" enthält:
# df_clean = df[~df['Headline_GER'].astype(str).str.contains('err', case=False, na=False)]

# 3. Ergebnis in eine neue Excel-Datei schreiben
output_path = "manuell_ergaenzt_bereinigt.xlsx"
df_clean.to_excel(output_path, index=False)

print(f"Fertig! Ursprüngliche Zeilen: {len(df)} | Bereinigte Zeilen: {len(df_clean)}")
```
