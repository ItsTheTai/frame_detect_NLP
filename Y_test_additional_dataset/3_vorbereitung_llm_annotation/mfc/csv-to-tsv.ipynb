```python
import csv
import os
import sys


def convert_csv_to_tsv(input_file, output_file=None):
    """Konvertiert eine CSV-Datei in eine TSV-Datei.

    Erkennt automatisch Komma (,) oder Semikolon (;) als Trennzeichen.
    """
    if not os.path.exists(input_file):
        print(f"Fehler: Die Datei '{input_file}' wurde nicht gefunden.")
        return

    if output_file is None:
        base, _ = os.path.splitext(input_file)
        output_file = base + ".tsv"

    try:
        with open(input_file, "r", encoding="utf-8-sig", newline="") as f_in:
            sample = f_in.read(4096)
            f_in.seek(0)

            try:
                dialect = csv.Sniffer().sniff(sample, delimiters=",;\t|")
                delimiter = dialect.delimiter
                print(f"Erkanntes Trennzeichen in der CSV: '{delimiter}'")
            except csv.Error:
                delimiter = ","
                print(
                    "Trennzeichen konnte nicht automatisch erkannt werden."
                    " Verwende Komma (',')."
                )

            reader = csv.reader(f_in, delimiter=delimiter)

            with open(output_file, "w", encoding="utf-8", newline="") as f_out:
                writer = csv.writer(
                    f_out, delimiter="\t", quoting=csv.QUOTE_MINIMAL
                )

                rows_count = 0
                for row in reader:
                    writer.writerow(row)
                    rows_count += 1

            print(
                f"Erfolg! {rows_count} Zeilen wurden in '{output_file}'"
                " gespeichert."
            )

    except Exception as e:
        print(f"Fehler bei der Konvertierung: {e}")


if __name__ == "__main__":
    DEFAULT_FILENAME = "mfc_dataset_headlines_full_LLM.csv"

    # Filtert Jupyter-Parameter heraus
    clean_args = [arg for arg in sys.argv[1:] if not arg.startswith("-")]

    if len(clean_args) > 0:
        input_csv = clean_args[0]
        output_tsv = clean_args[1] if len(clean_args) > 1 else None
        convert_csv_to_tsv(input_csv, output_tsv)
    else:
        # Wenn die Datei im selben Ordner existiert, direkt umwandeln
        if os.path.exists(DEFAULT_FILENAME):
            print(f"Verwende Standard-Datei: '{DEFAULT_FILENAME}'")
            convert_csv_to_tsv(DEFAULT_FILENAME)
        else:
            user_input = input(
                f"Bitte Dateipfad eingeben (Standard: {DEFAULT_FILENAME}): "
            ).strip()
            filename = user_input if user_input else DEFAULT_FILENAME
            convert_csv_to_tsv(filename)
```
