```python
import pandas as pd

# 1. Load the Excel files
df_ground_truth = pd.read_excel('ground_truth_full.xlsx')
df_filter = pd.read_excel('03_finale_titelannotationen.xlsx')

# 2. Extract valid IDs from the second file as a set for quick lookup
valid_ids = df_filter['article id']

# 3. Filter ground_truth to only keep rows matching those IDs
filtered_df = df_ground_truth[df_ground_truth['Filename'].isin(valid_ids)]

# 4. Save the filtered result to a new Excel file
filtered_df.to_excel('ground_truth_filtered.xlsx', index=False)
```
