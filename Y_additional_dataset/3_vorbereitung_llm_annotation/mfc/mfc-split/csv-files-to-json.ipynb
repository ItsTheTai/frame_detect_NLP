```python
import glob
import pandas as pd

# Alle passenden CSV-Dateien suchen und nach Dateinummer sortieren
csv_files = sorted(
    glob.glob("mfc_dataset_headlines_part_*.csv"),
    key=lambda x: int(x.rsplit("_", 1)[-1].split(".")[0]),
)

for csv_file in csv_files:
    # CSV-Datei einlesen (Semikolon-getrennt, UTF-8-SIG)
    df = pd.read_csv(csv_file, sep=";", encoding="utf-8-sig")

    # Dateiname für JSON bestimmen (z. B. mfc_dataset_headlines_part_1.json)
    json_file = csv_file.rsplit(".", 1)[0] + ".json"

    # Als JSON-Liste von Objekten speichern
    df.to_json(json_file, orient="records", force_ascii=False, indent=4)
    print(f"Konvertiert: {csv_file} -> {json_file}")
```
